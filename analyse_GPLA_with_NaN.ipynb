Nettoyage texte

```python
import re
import unicodedata
```

```python
def clean_text(text):
    if pd.isna(text):
        return text
    
    # Forcer le type string
    text = str(text)
    
    # Minuscules
    text = text.lower()
    
    # Suppression des accents
    text = unicodedata.normalize("NFD", text)
    text = "".join(c for c in text if unicodedata.category(c) != "Mn")
    
    # Suppression ponctuation / caractères spéciaux
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    
    # Nettoyage espaces
    text = re.sub(r"\s+", " ", text).strip()
    
    return text
```

Extraction des elements ayant potentiellement de la chalconatronite en se basant sur mots cles (poids 1-2-3)

Logique de scoring=>
Pour chaque enregistrement, on parcourt les colonnes textuelles et on cumule les points :
score = Σ (nombre d'occurrences du mot-clé × son poids)
Puis on applique un seuil de décision (ex: score ≥ 3 → l'élément a la caractéristique "x").

```python
df_cuivre
```

```python
# Colonnes textuelles à analyser
TEXT_COLUMNS = ["Matière_element", "Description_analytique","Intervention_requise","Degradation","Techniques"]  


keywords = {
    3: ["chalco","chalconatronite","chalconatronitre"], #poids fort
    2: ["sesqui","sesquicarbonate","sesquicarbonates","Chlorure","Dégradation","Altération","Corrosion localisée","Corrosion verte","Corrosion active","Produit de corrosion bleu-vert","bleu-vert","Pulvérulent","Pulvérulence","Récurrence cyclique","Corrosion pulvérulente","Sels verts"],   # poids moyen
    1: ["Archéologie","Sculpture","Croyance","Croyances-coutumes","Statuette","Statue","Amulette","Monnaie","Basse époque","Époque ptolémaïque","Antiquité","Égypte","Moyen Orient","Pratique votive","Fonte à la cire perdue","Coulé","Cire perdue","Stabilisé","Oxydes d'argent","Protection de la surface : Paraloïd B72 à 5%"
,"Paraloïd B72 à 5%","oxydation","stabilisation chimique"],# poids faible
}
```

```python
keywords = {
    k: [clean_text(mot) for mot in liste]
    for k, liste in keywords.items()
}
```

```python
keywords
```

```python
# Seuil de détection 
#   - 1 mot de poids 2                   → score = 2  
#   - 2 mots de poids 1                  → score = 2  
#   - 1 mot poids 2 + 1 mot poids 1      → score = 3  
SEUIL = 2
```

```python
# Fonction de scoring 
def calculer_score(row):
    # Concaténer toutes les colonnes textuelles en une seule chaîne
    texte = " ".join(
        str(row[col]).lower() for col in TEXT_COLUMNS if col in row.index
    )
    
    score = 0
    details = []
    
    for poids, mots in keywords.items():
        for mot in mots:
            pattern = r'\b' + re.escape(mot.lower()) + r'\b'
            if re.search(pattern, texte):
                score += poids
                details.append(f"'{mot}' (poids {poids})")
    
    return score, "; ".join(details)
```

```python
# Application sur le DataFrame 
df_cuivre[["score_chalco", "mots_detectes"]] = df_cuivre.apply(
    lambda row: pd.Series(calculer_score(row)), axis=1
)
```

```python
# Détection finale 
df_cuivre["est_chalco"] = df_cuivre["score_chalco"] >= SEUIL
```

```python
#Résultats 
print(f"Éléments détectés comme 'chalconatronite' : {df_cuivre['est_chalco'].sum()} / {len(df_cuivre)}")
print(f"Non détectés             : {(~df_cuivre['est_chalco']).sum()}")
print(df_cuivre[df_cuivre["est_chalco"]][["score_chalco", "mots_detectes"]].head(10))
```
